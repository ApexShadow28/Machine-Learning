1) Provisioning and running machine learning workflows in GCP Colab Enterprise involves dependencies on identity, quota, and billing that do not exist in regular free-tier notebooks, necessitating an active Google Cloud Project associated with education credits and with enabled Vertex AI APIs. Furthermore, when doing work inside the managed cloud containers, file paths become isolated, resulting in the failure of file operations when the scripts are executed outside the defined working directory of the container. Resource lifecycle management is another issue due to the continuous consumption of credits by active instances and multi-hour delays in the cloud billing dashboards.

2) The existence of the implicit missing values was discovered with the help of descriptive statistics (df.describe()) that showed impossible minimums of -200 in all sensor columns, while the standard detection of missing values indicated the absence of any missing values. These sentinel values were changed to np.nan, which allowed detection of such missing values, with more than 8,400 in NMHC(GT). Missing values were dealt with the help of Scikit-Learn's SimpleImputer with the use of the median strategy. It helped to stabilize further models because of possible outliers in air quality measurements.


3) The Scikit-Learn ColumnTransformer and Pipeline improve feature preparation by cleanly separating heterogenous data types into either scaling or one-hot encoding based on whether they consist of continuous sensor telemetry data or categorical timestamps respectively in a single process. Most importantly, wrapping the above processes in a pipeline makes sure that there is no data leakage by computing the imputation medians and scaling values $\mu, \sigma$ only from the training set using the .fit() method and then using them on the test set using .transform().

4) The StandardScaler ensures zero mean and unit variance for the feature vectors and brings the loss surface in gradient descent algorithms like SGDClassifier into an aspherical shape in order to speed up the optimization process. However, MinMaxScaler squeezes the features into the range of [0, 1] and, hence, becomes very sensitive to spikes in environmental pollution and reduces the effect of gradient descent on the squeezed sensor values. Hence, although both scalers provide similar accuracy at the end of the process after proper tuning, the former is more stable and requires fewer iterations to converge.

5) While both closed-form and batch approaches analyze the whole training dataset at once to get accurate gradients and matrix inverses at the expense of significant computational and space complexity, the SGDClassifier learns from the dataset using iterative analysis with mini-batches or single examples, thus making the approach memory-efficient on large or streaming datasets but needing proper setting of the learning rate and number of epochs. Using the loss="log_loss" parameter for configuring the SGDClassifier makes it provide continuous calibrated probabilities via .predict_proba() instead of decision thresholds, making it possible for developers to determine certainty for the pollution classes.

6) With an increase in the regularization parameter, Ridge Regression progressively reduced the coefficients towards zero to counteract multicollinearity between sensors without reducing any features and maintaining maximum accuracy, while Lasso reduced unnecessary coefficients to zero, leading to more sparsity by eliminating 406 to 426 features from a total of 427 columns. Environmentally, it is evident from the above findings that while sensor telemetry gives us maximum precision, a response from specific sensors is most useful for predicting Benzene levels. This finding has practical applications where environmental stations may use feature selection in Lasso to get rid of costly sensors.
