When loading external datasets such as Telco Churn, columns where numbers are missing tend to use empty strings ' ' rather than NaN for identifying missing entries. As a result, Pandas will automatically convert columns with missing numeric values like TotalCharges to string objects rather than using their true type, which requires manual transformation to np.nan in order to preprocess them.

The columns were checked by using df.isnull().sum(), after replacing the implicit whitespace string. There were 11 NaN in the TotalCharges column due to the existence of new customers with a tenure of zero months. Imputation of numerical features was done by using median (SimpleImputer(strategy='median')), and mode (strategy='most_frequent') for categorical features.

ColumnTransformer and Pipeline of Scikit-Learn combine transformations, encodings, and estimators in one unit. While validating the model, Pipeline.fit() finds parameters of transformations (for instance, mean and std for StandardScaler) only from the training data subset. Such parameters are used in .predict() without recalculations from the test/validate dataset to avoid data leakage.

StandardScaler standardizes features by making them have a mean of 0 and unit variance, whereas MinMaxScaler scales down the features to the range [0, 1]. In stochastic gradient descent (SGDClassifier), StandardScaler is useful for making the gradient smoother and converging the weights faster where numeric features have a normal distribution or have outliers.

Setting up the SGDClassifier with loss set to 'log_loss' optimizes the logistic regression with an iterative mini-batch stochastic gradient descent, not exact closed-form optimization techniques. Thus, this algorithm generates class probability estimates (predict_proba) instead of hard binary predictions. Assessing churn via ROC-AUC assesses how well the classifier identifies true positives for all probability scores (0.0 to 1.0).

Positive Coefficients (High Risk of Attrition): Monthly contracts, fiber optic connection of Internet services, and high MonthlyCharges are among the positive coefficients. Uncommitted customers with a high monthly rate pose the greatest risk of attrition.

Negative Coefficients (Low Risk of Attrition / Improves Customer Retention): Long customer tenure, two-year contracts, and automated payments have negative coefficients. This indicates that the customers’ long tenure and automatic payments minimize the risk of attrition.
