Variables such as insulin (48.70% zeros) and skin_thickness (29.56% zeros) can not have a biological value of 0 for live patients, so are latent missing data. Removing these rows would remove almost half the data and cause massive bias. By replacing 0 with np.nan and using median imputation we can impute the data without losing any information from the 768 patients and avoid extreme values of insulin.

In Hard Voting the class predictions are aggregated using a basic majority voting system (i.e if 2 out of 3 models predict class x then it will be assigned to class x). In Soft Voting the probability of each class prediction is averaged, meaning that more accurate models will have more impact on the final prediction. Despite achieving greater accuracy (73.38% compared to 72.08%), we must ensure that the probabilities are well calibrated so that our model doesn’t become too over-confident.

As we are using bootstrap sampling with replacement, approximately 36.8% of data will not be included in each tree. This means that the data acts as an internal validation set and we can calculate how well our model generalises without having to split into cross-validation. We can see that our OOB score was very similar to our accuracy on the test data (0.7557 and 0.7208) which means our model is well validated.

Using Standard Bagging forces all features to be used at each split in the trees. This results in the most important feature such as glucose appearing at the top of nearly all trees. Random Forests forces each tree to only select from a random subset of features ($\sqrt{p}$). This creates a wide variety of uncorrelated trees and reduces the overall variance. This resulted in a higher accuracy when testing our model (Random Forest achieved 77.27% compared to 72.08% in Bagging).

AdaBoost sequentially fits decision stumps to increase the weights of incorrectly classified patients. As shown in the plot above, our training accuracy increased up to 79.80%. However our test accuracy had plateaued at around 74.68% meaning that there was some overfitting. This happened because later trees became too tailored to noisy outliers. This can be reduced by reducing our learning_rate, reducing the number of estimators or using early stopping.
