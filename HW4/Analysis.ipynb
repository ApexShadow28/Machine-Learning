The dataset combines large-scale vitals like cholesterol with small-scale categories like chest pain. Support Vector Machines rely on distance calculations, so unscaled large numbers would completely drown out the categorical features. In contrast, Decision Trees split one feature at a time based on rank order, making them completely unaffected by scaling.

Hidden missing values were entered as ? strings instead of blank cells, hiding 4 missing entries in ca and 2 in thal. We used median imputation for continuous vitals because it resists extreme outliers like high blood pressure or cholesterol. We used mode imputation for categorical tests like ca and thal to avoid inventing invalid decimal categories.

The Support Vector Classifier uses an RBF kernel to draw a smooth, curved boundary in high-dimensional space, outperforming the tree with an accuracy of 89% and an ROC-AUC of 0.9643. However, the SVC is a "black box," while the Decision Tree provides a step-by-step flowchart that doctors can easily read, audit, and trust.

The parameter C controls how strictly the model avoids training errors, where a higher C risks overfitting and a lower C risks underfitting. The parameter gamma sets how tightly the boundary wraps around individual data points, where a high gamma creates complex, overfitted margins. GridSearchCV chose C = 10 and gamma = 0.001, finding a broad, smooth boundary that achieved a cross-validation ROC-AUC of 0.8947 without memorizing training noise.
