Logistic Regression after PCA dimensionality reduction gave a test accuracy score of 97.50% which is equal to the baseline on raw features and the Soft Voting Ensemble model score (97.50%) and just outperforms the RBF Kernel SVM (95.00%). The use of non-linear kernels and ensemble averaging did not improve performance greatly because the dataset contains standardised and centred facial poses meaning that there was a high level of linear separability between classes in the PCA latent space. Furthermore, all the classical classifiers suffered from an intrinsic limitation on image data as they view images as 1D arrays. This means there is no notion of spatial information or translation invariance. Therefore, if a subject moves their head slightly, changes expression or moves just a few pixels then the distances between the 1D array will change significantly however the true semantic label remains the same.

As seen in Task 8, PCA significantly improves training efficiency. Logistic Regression can be trained in much less time using PCA. The Logistic Regression time was reduced from 15.263 seconds to 1.020 seconds, however no reduction in accuracy is seen. In this case, global linear projections are not enough for the modern computer vision tasks. The PCA method is global linear and cannot account for real world vision problems. PCA will not be able to account for random camera angles, varying backgrounds, occlusions and extreme nonlinear changes in lighting. To combat this issue modern computer vision uses deep learning methods such as Convolutional Neural Networks (CNNs) and Vision Transformers (ViTs). Deep learning uses local filters to account for translation invariance. This allows the network to learn a hierarchical representation of the features within the data. This means that the network can learn low level features first, then mid level features such as parts of a face and finally high level invariant features.
